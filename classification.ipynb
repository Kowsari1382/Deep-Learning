```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense, Flatten
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import normalize
from sklearn.datasets import load_iris
```

```python
data = load_iris()
x = data.data
y = data.target
```

```python
new_y = keras.utils.to_categorical(y, num_classes= 3)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, new_y, test_size= 0.2, random_state=42)
```

```python
x_train = normalize(x_train)
x_test = normalize(x_test)
```

```python
model = keras.Sequential(
    [
        Dense(100, activation="relu"),
        Dense(100, activation="relu"),
        Dense(100, activation="relu"),
        Dense(100, activation="relu"),
        Dense(3, activation="linear"),
    ]
)
```

```python
model.compile(
    loss=keras.losses.CategoricalCrossentropy(from_logits=True),
    optimizer="adam",
    metrics=["accuracy"],
)
```

```python
history = model.fit(x_train, y_train, epochs= 100, batch_size= 32)
```

```python
score1 = model.evaluate(x_test, y_test)
score2 = model.evaluate(x_train, y_train)
score1, score2
```

```python
probality_model = keras.Sequential([
    model,
    keras.layers.Softmax()
])
```

```python
ypred = probality_model.predict(x_test)
```

```python
np.mean(np.argmax(y_test, axis=-1) == np.argmax(ypred, axis=-1))
```
